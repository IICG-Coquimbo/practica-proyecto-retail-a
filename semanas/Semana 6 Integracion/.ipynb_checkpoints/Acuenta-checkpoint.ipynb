```python
# --- SCRAPER CANASTA BÁSICA ACUENTA.CL - VERSIÓN PAGINACIÓN AVANZADA ---
import os
import time
import random
import re
from pymongo import MongoClient
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException, StaleElementReferenceException

os.system("pkill -9 chrome")
os.system("pkill -9 chromedriver")
print("🧹 Scraper Canasta Básica iniciado")

# VARIABLES
URL_CANASTA = "https://www.acuenta.cl/ca/despensa/05"
NOMBRE_GRUPO = "Canasta Básica Acuenta"
datos_finales = []
pagina_actual = 1

# Configuración anti-detección
options = Options()
options.binary_location = "/usr/bin/google-chrome"
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/123.0.0.0 Safari/537.36")

driver = None
try:
    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
        "source": "Object.defineProperty(navigator, 'webdriver', {get: () => })"
    })
    print("🚀 Navegador listo")

    # 🔄 SISTEMA DE PAGINACIÓN AVANZADO
    while True:
        print(f"\n📄 === PÁGINA {pagina_actual} ===")
        print(f"🛍️ Cargando: {URL_CANASTA}")
        
        driver.get(URL_CANASTA)
        time.sleep(random.uniform(8, 12))

        # Cookies
        try:
            cookie_btn = WebDriverWait(driver, 5).until(
                EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Aceptar') or contains(text(), 'Acepto') or contains(@class, 'cookie')]"))
            )
            cookie_btn.click()
            print("🍪 Cookies aceptadas")
            time.sleep(2)
        except:
            print("ℹ️ Sin cookies")

        # 📜 Scroll agresivo por página
        print("📜 Scroll para cargar productos...")
        last_height = driver.execute_script("return document.body.scrollHeight")
        scroll_count = 0
        
        while scroll_count < 15:
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
            time.sleep(random.uniform(1.5, 3))
            new_height = driver.execute_script("return document.body.scrollHeight")
            
            if new_height == last_height:
                time.sleep(2)
                if driver.execute_script("return document.body.scrollHeight") == new_height:
                    break
            last_height = new_height
            scroll_count += 1
        
        time.sleep(3)

        # 🔍 SELECTORES OPTIMIZADOS
        print("🔍 Buscando productos...")
        
        selectores_canasta = [
            "div[class*='product']", "div[class*='ProductCard']",
            "div[class*='product-card']", "div[class*='card']",
            ".product-item", "[data-testid*='product']",
            "article[class*='product']", ".grid-item", ".product-grid-item"
        ]
        
        bloques = []
        for selector in selectores_canasta:
            try:
                bloques = driver.find_elements(By.CSS_SELECTOR, selector)
                if len(bloques) > 2:
                    print(f"✅ {len(bloques)} productos con '{selector}'")
                    break
            except:
                continue

        # Fallback por precios
        if not bloques or len(bloques) < 3:
            print("🔍 Fallback: buscando por PRECIOS...")
            bloques = driver.find_elements(By.XPATH, "//*[contains(text(), '$')]/ancestor::div[1|2|3]")
            bloques = list(set(bloques))[:50]
            print(f"✅ {len(bloques)} productos por precio")

        if not bloques:
            print("⚠️ No se encontraron productos en esta página")
            break

        # 📦 EXTRACCIÓN MEJORADA CON TODAS LAS ETIQUETAS
        print(f"📦 Procesando {len(bloques)} productos...")
        productos_pagina = 0
        
        for i, bloque in enumerate(bloques[:50]):
            try:
                # 🏷️ NOMBRE
                nombre = None
                selectores_nombre = [
                    "[class*='name']", "[class*='title']", "h3", "h2", "h4",
                    ".product-name", ".titulo", "[data-testid*='name']", ".product-title"
                ]
                
                for sel in selectores_nombre:
                    try:
                        elem = bloque.find_element(By.CSS_SELECTOR, sel)
                        nombre_cand = elem.text.strip()
                        if len(nombre_cand) > 3 and len(nombre_cand) < 120:
                            nombre = nombre_cand
                            break
                    except:
                        continue
                
                # Fallback nombre
                if not nombre:
                    texto = bloque.text.strip()
                    lineas = [l.strip() for l in texto.split('\n') if l.strip()]
                    for linea in lineas:
                        if '$' not in linea and len(linea) > 3 and len(linea) < 100:
                            nombre = linea
                            break

                # 💰 PRECIO
                precio = None
                selectores_precio = [
                    "[class*='price']", "[class*='precio']", ".price",
                    "span[class*='price']", "[data-testid*='price']",
                    ".precio-final", ".current-price"
                ]
                
                for sel in selectores_precio:
                    try:
                        elem = bloque.find_element(By.CSS_SELECTOR, sel)
                        precio_cand = elem.text.strip()
                        if '$' in precio_cand:
                            precio = precio_cand
                            break
                    except:
                        continue
                
                # Fallback precio
                if not precio:
                    match = re.search(r'\$[\d,\.\s]+', bloque.text)
                    precio = match.group() if match else None

                # 🏪 MARCA
                marca = None
                selectores_marca = [
                    "[class*='brand']", "[class*='marca']", ".brand-name",
                    "[data-testid*='brand']"
                ]
                
                for sel in selectores_marca:
                    try:
                        elem = bloque.find_element(By.CSS_SELECTOR, sel)
                        marca_cand = elem.text.strip()
                        if len(marca_cand) > 1 and len(marca_cand) < 30:
                            marca = marca_cand.upper()
                            break
                    except:
                        continue

                # ✅ GUARDAR CON TODAS LAS ETIQUETAS
                if nombre and precio:
                    datos_finales.append({
                        "nombre": nombre[:100],
                        "precio": precio,
                        "precio_limpio": 0.0,
                        "marca": marca or "SIN MARCA",
                        "supermercado": "ACUENTA",
                        "fecha_scraping": time.strftime("%Y-%m-%d %H:%M:%S"),
                        "grupo": NOMBRE_GRUPO,
                        "url": URL_CANASTA,
                        "pagina": pagina_actual,
                        "posicion": len(datos_finales) + 1,
                        "categoria": "Canasta Básica"
                    })
                    productos_pagina += 1
                    print(f"✅ [{productos_pagina:2d}] {nombre[:40]:<40} ${precio} | {marca or ''}")

            except Exception as e:
                continue

        print(f"🎉 Página {pagina_actual}: {productos_pagina} productos")

        # 🔎 DETECTAR SIGUIENTE PÁGINA
        print("🔎 Buscando siguiente página...")
        selectores_paginacion = [
            "a[aria-label*='siguiente']", "a[aria-label*='next']",
            ".pagination-next", ".next-page", "a[class*='next']",
            "button[class*='next']", ".page-next",
            "//a[contains(text(), 'Siguiente') or contains(text(), '→')]",
            "//button[contains(text(), 'Siguiente') or contains(text(), '→')]"
        ]
        
        siguiente_encontrado = False
        for selector in selectores_paginacion:
            try:
                if "//" in selector:
                    elementos = driver.find_elements(By.XPATH, selector)
                else:
                    elementos = driver.find_elements(By.CSS_SELECTOR, selector)
                
                for elem in elementos:
                    if elem.is_enabled() and elem.is_displayed():
                        driver.execute_script("arguments[0].scrollIntoView(true);", elem)
                        time.sleep(1)
                        elem.click()
                        print(f"✅ → PÁGINA {pagina_actual + 1} (selector: {selector[:30]})")
                        pagina_actual += 1
                        siguiente_encontrado = True
                        time.sleep(random.uniform(3, 5))
                        break
                if siguiente_encontrado:
                    break
            except (StaleElementReferenceException, TimeoutException):
                continue

        # 🛑 DETENER SI NO HAY MÁS PÁGINAS
        if not siguiente_encontrado or productos_pagina < 3:
            print("🛑 No hay más páginas")
            break

        # ⏸️ PAUSA ENTRE PÁGINAS
        time.sleep(random.uniform(5, 8))

    print(f"\n🎉 TOTAL: {len(datos_finales)} productos de {pagina_actual} páginas")

except Exception as e:
    print(f"❌ Error: {e}")
finally:
    if driver:
        driver.quit()

# 💰 LIMPIEZA PROFESIONAL
if datos_finales:
    print("\n💰 Limpiando precios...")
    precios_limpios = []
    
    for d in datos_finales:
        # Patrón pesos chilenos
        patron = r'\$([0-9]{1,3}(?:\.[0-9]{3})*(?:,\d+)?)'
        match = re.search(patron, d["precio"])
        
        if match:
            precio_str = match.group(1).replace('.', '').replace(',', '.')
            try:
                precio = float(precio_str)
                if precio > 0:
                    d["precio_limpio"] = precio
                    precios_limpios.append(precio)
            except:
                continue

    # 📊 TABLA COMPLETA CON TODAS LAS ETIQUETAS
    print("\n" + "═"*120)
    print("🛒  CANASTA BÁSICA ACUENTA.CL - COMPLETA")
    print(f"🔗 {URL_CANASTA} | 📄 {pagina_actual} páginas")
    print("═"*120)
    print(f"{'PRODUCTO':<45} {'MARCA':<12} {'PRECIO':>10} {'PÁG':>4}")
    print("─"*120)
    
    for d in datos_finales:
        if d["precio_limpio"] > 0:
            print(f"{d['nombre'][:45]:<45} {d['marca'][:12]:<12} ${d['precio_limpio']:>9,.0f} {d['pagina']:>4}")
    
    # 📈 ESTADÍSTICAS
    if precios_limpios:
        total = sum(precios_limpios)
        promedio = total / len(precios_limpios)
        
        print("─"*120)
        print(f"{'TOTAL PRODUCTOS:':<45} {len(precios_limpios):>12} {'PROMEDIO:':<12} ${promedio:>9,.0f}")
        print(f"{'COSTO TOTAL CANASTA:':<45} ${total:>9,.0f} {'MÁXIMO:':<12} ${max(precios_limpios):>9,.0f}")
        print(f"{'SUPERMERCADO:':<45} {datos_finales[0]['supermercado']:>12} {'FECHA:':<12} {datos_finales[0]['fecha_scraping']}")
        print("═"*120)

        # 💾 EXPORTACIONES
        import json
        
        # JSON completo
        with open("canasta_basica_acuenta_COMPLETA.json", "w", encoding="utf-8") as f:
            json.dump(datos_finales, f, ensure_ascii=False, indent=2)
        print("💾 ✅ canasta_basica_acuenta_COMPLETA.json")
        
        # Resumen TXT
        with open("resumen_canasta_basica_COMPLETO.txt", "w", encoding="utf-8") as f:
            f.write(f"CANASTA BÁSICA ACUENTA.CL\n")
            f.write(f"URL: {URL_CANASTA}\n")
            f.write(f"Páginas: {pagina_actual} | Fecha: {time.strftime('%Y-%m-%d %H:%M:%S')}\n\n")
            f.write(f"TOTAL: {len(precios_limpios)} productos | COSTO: ${total:,.0f}\n")
            f.write(f"PROMEDIO: ${promedio:,.0f} | Supermercado: ACUENTA\n\n")
            f.write("DETALLE:\n" + "─"*80 + "\n")
            for d in datos_finales:
                if d["precio_limpio"] > 0:
                    f.write(f"${d['precio_limpio']:>9,.0f} | {d['marca']:<12} | {d['nombre']}\n")
        
        print("💾 ✅ resumen_canasta_basica_COMPLETO.txt")
        
        # MongoDB
        try:
            client = MongoClient("mongodb", 27017, serverSelectionTimeoutMS=5000)
            db = client["Canasta_db"]
            db["Canasta_Basica_Acuenta_Completa"].insert_many(datos_finales)
            print("💾 ✅ MongoDB COMPLETO")
        except:
            print("⚠️ Sin MongoDB")

else:
    print("❌ No se encontraron productos")
```

```python
# --- SCRAPER 15 PÁGINAS - ULTRA ESTABLE CON RECONEXIÓN ---
import os
import time
import random
import re
import json
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException, WebDriverException

os.system("pkill -9 chrome")
os.system("pkill -9 chromedriver")
print("🧹 SCRAPER ULTRA-ESTABLE - 15 PÁGINAS CON RECONEXIÓN")

URL_BASE = "https://www.acuenta.cl/ca/despensa/05"
MAX_PAGINAS = 15
datos_finales = []
productos_vistos = set()

def crear_driver():
    """🔧 Crea driver estable"""
    options = Options()
    options.binary_location = "/usr/bin/google-chrome"
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option('useAutomationExtension', False)
    options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36")
    
    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
        "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined})"
    })
    return driver

# 🌈 CATEGORÍAS
CATEGORIAS = {
    'arroz': 'Arroz', 'fideo': 'Pastas', 'pasta': 'Pastas', 'espagueti': 'Pastas',
    'aceite': 'Aceites', 'mayonesa': 'Salsas', 'ketchup': 'Salsas', 'salsa': 'Salsas',
    'atún': 'Conservas', 'jurel': 'Conservas', 'caballa': 'Conservas', 'azúcar': 'Azúcares', 
    'harina': 'Harinas', 'porotos': 'Legumbres', 'lentejas': 'Legumbres'
}

def extraer_producto(bloque):
    """🎯 Extracción AGRESIVA - SIEMPRE funciona"""
    try:
        texto_completo = bloque.text.strip()
        if len(texto_completo) < 15:
            return None
            
        # 💰 PRECIO - MÁS FLEXIBLE
        patron_precio = r'\$[\d,\.\s]+(?:\$[\d,\.\s]+\$)?'
        precio_match = re.search(patron_precio, texto_completo)
        if not precio_match:
            return None
        precio = precio_match.group()
        
        # 📝 NOMBRE - CUALQUIER TEXTO CON >3 CARACTERES
        lineas = texto_completo.split('\n')
        candidatos = []
        for linea in lineas:
            linea_limpia = linea.strip()
            if len(linea_limpia) > 3 and '$' not in linea_limpia and not linea_limpia.isdigit():
                candidatos.append(linea_limpia)
        
        if not candidatos:
            return None
            
        nombre_completo = max(candidatos, key=len)
        
        # 🏷️ MARCA
        marca = extraer_marca(nombre_completo)
        nombre_limpio = limpiar_nombre(nombre_completo, marca)
        
        # ID único
        identificador = f"{nombre_limpio[:60].lower()}_{marca.lower()}"
        if identificador in productos_vistos:
            return None
        productos_vistos.add(identificador)
        
        return {
            "nombre": nombre_limpio[:100],
            "nombre_completo": nombre_completo[:120],
            "precio": precio,
            "marca": marca,
            "categoria": detectar_categoria(nombre_limpio.lower()),
            "pagina": pagina_actual
        }
    except:
        return None

def extraer_marca(nombre):
    marcas = ['LUCCHETTI','TUCAPEL','IANSA','CAROZZI','MIRAFLORES','HELLMANNS','LIDER','ACUENTA','SELECTA','LA ROMA','NETO','KRAFT','MAGGI']
    for marca in marcas:
        if marca.upper() in nombre.upper():
            return marca
    return "GENÉRICA"

def limpiar_nombre(nombre, marca):
    return re.sub(rf'\b{re.escape(marca)}\b\s*$', '', nombre, flags=re.IGNORECASE).strip()

def detectar_categoria(nombre):
    nombre_lower = nombre.lower()
    for palabra, cat in CATEGORIAS.items():
        if palabra in nombre_lower:
            return cat
    return "Despensa"

# 🚀 SCRAPING PRINCIPAL CON RECONEXIÓN
driver = None
for pagina_actual in range(1, MAX_PAGINAS + 1):
    try:
        if driver is None or driver.session_id is None:
            print(f"\n🔄 RECONEXIÓN PÁGINA {pagina_actual}")
            if driver:
                driver.quit()
            driver = crear_driver()
            time.sleep(3)
        
        url_pagina = URL_BASE if pagina_actual == 1 else f"{URL_BASE}?currentPage={pagina_actual}"
        print(f"\n📄 PÁGINA {pagina_actual:2d}/15 ({pagina_actual/MAX_PAGINAS*100:>5.1f}%)")
        
        driver.get(url_pagina)
        time.sleep(8)

        # Cookies
        try:
            WebDriverWait(driver, 5).until(EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Aceptar') or contains(text(), 'Acepto')]"))).click()
        except:
            pass

        # SCROLL AGRESIVO
        for _ in range(25):
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
            time.sleep(1.2)
        time.sleep(5)

        # 🔍 TODOS LOS SELECTORES POSIBLES
        selectores = [
            "div[class*='product']",
            "div[class*='ProductCard']",
            "div[class*='product-card']",
            "article",
            ".product-item"
        ]
        
        todos_bloques = []
        for selector in selectores:
            try:
                bloques = driver.find_elements(By.CSS_SELECTOR, selector)
                todos_bloques.extend(bloques)
                print(f"   🔍 {selector}: {len(bloques)}")
            except:
                pass
        
        todos_bloques = list(set(todos_bloques))  # Eliminar duplicados
        
        print(f"   ✅ {len(todos_bloques)} bloques únicos")

        # EXTRAER
        productos_esta_pagina = 0
        for bloque in todos_bloques:
            producto = extraer_producto(bloque)
            if producto:
                producto["supermercado"] = "ACUENTA"
                producto["fecha_scraping"] = time.strftime("%Y-%m-%d %H:%M:%S")
                producto["url"] = url_pagina
                producto["posicion"] = len(datos_finales) + 1
                datos_finales.append(producto)
                productos_esta_pagina += 1
        
        print(f"   🎉 +{productos_esta_pagina:3d} productos | TOTAL: {len(datos_finales):4d}")

    except WebDriverException as e:
        print(f"   ❌ WebDriver error: {e}")
        driver = None  # Forzar reconexión
        time.sleep(5)
        continue
    except Exception as e:
        print(f"   ⚠️ Error página {pagina_actual}: {e}")
        time.sleep(3)
        continue

print(f"\n🎉 ¡SCRAPE FINALIZADO! {len(datos_finales)} productos de 15 páginas")

# 💰 LIMPIEZA
precios_limpios = []
for d in datos_finales:
    match = re.search(r'\$([0-9]{1,3}(?:\.[0-9]{3})*)', d["precio"])
    if match:
        try:
            d["precio_limpio"] = float(match.group(1).replace('.', ''))
            precios_limpios.append(d["precio_limpio"])
        except:
            pass

# 📊 TABLA
print("\n" + "="*140)
print(f"{'PRODUCTOS TOTALES':<45} {len(datos_finales):>5} | {'COSTO TOTAL':<15} ${sum(precios_limpios):>10,.0f}")
print("="*140)

# TOP 20 + ÚLTIMOS 10
for i, d in enumerate(datos_finales[:20], 1):
    precio = d.get("precio_limpio", 0)
    print(f"{i:>3}. {d['nombre'][:45]:<45} {d['marca']:<12} ${precio:>8,.0f} ({d['pagina']})")

print("..." + "."*60)
for i, d in enumerate(datos_finales[-10:], len(datos_finales)-9):
    precio = d.get("precio_limpio", 0)
    print(f"{i:>3}. {d['nombre'][:45]:<45} {d['marca']:<12} ${precio:>8,.0f} ({d['pagina']})")

# 💾
with open("canasta_15p_estable.json", "w", encoding="utf-8") as f:
    json.dump(datos_finales, f, ensure_ascii=False, indent=2)
print("\n💾 ✅ canasta_15p_estable.json")

print("\n🎉 ¡15 PÁGINAS ESTABLES! ✅")
```

```python
import os
import time
import re
import json
import statistics
from collections import defaultdict
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException, WebDriverException

# 🧹 Limpieza de sistema
os.system("pkill -9 chrome")
os.system("pkill -9 chromedriver")

print("🧹 INICIANDO EXTRACCIÓN ULTRA ESTABLE - ACUENTA PROFESIONAL v2.0")

URL_BASE = "https://www.acuenta.cl/ca/despensa/05"
MAX_PAGINAS = 15
datos_finales = []
productos_vistos = set()
reconexiones = 0

def crear_driver():
    """🔧 Driver ultra estable con anti-detección"""
    options = Options()
    options.binary_location = "/usr/bin/google-chrome"
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option('useAutomationExtension', False)
    options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    # HEADLESS OPCIONAL - descomenta para modo invisible
    # options.add_argument("--headless")
    
    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
        "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined})"
    })
    return driver

# 🌈 CATEGORÍAS COMPLETAS
CATEGORIAS = {
    'arroz': 'Arroz', 'fideo': 'Pastas', 'pasta': 'Pastas', 'espagueti': 'Pastas', 'tallarines': 'Pastas',
    'fettucine': 'Pastas', 'corbatas': 'Pastas', 'espirales': 'Pastas', 'macarrones': 'Pastas',
    'aceite': 'Aceites', 'mayonesa': 'Salsas', 'ketchup': 'Salsas', 'salsa': 'Salsas', 'mostaza': 'Salsas',
    'atún': 'Conservas', 'jurel': 'Conservas', 'caballa': 'Conservas', 'chorito': 'Conservas',
    'azúcar': 'Azúcares', 'harina': 'Harinas', 'porotos': 'Legumbres', 'lentejas': 'Legumbres', 'garbanzos': 'Legumbres',
    'sal': 'Condimentos', 'orégano': 'Condimentos', 'pimentón': 'Condimentos', 'sopa': 'Instantáneos'
}

def extraer_marca(nombre):
    """🏪 Detección mejorada de marcas"""
    marcas = ['LUCCHETTI','TUCAPEL','IANSA','CAROZZI','MIRAFLORES','HELLMANNS','LIDER','ACUENTA','SELECTA','LA ROMA','NETO','KRAFT','MAGGI','TRAVERSO','HOFFMANN','GOURMET','LOBOS','VIVO','BANQUETE']
    nombre_upper = nombre.upper()
    for marca in marcas:
        if marca in nombre_upper:
            return marca
    return "GENÉRICA"

def detectar_categoria(nombre):
    """🏷️ Categorización inteligente"""
    nombre_lower = nombre.lower()
    for palabra, cat in CATEGORIAS.items():
        if palabra in nombre_lower:
            return cat
    return "Despensa General"

def extraer_precio(texto):
    """💰 Extracción robusta de precios"""
    match = re.search(r'\$([0-9]{1,3}(?:\.[0-9]{3})*(?:,\d+)?)', texto)
    if match:
        try:
            return float(match.group(1).replace('.', '').replace(',', '.'))
        except:
            pass
    return 0.0

def limpiar_para_id(nombre_raw):
    """🧹 Limpiar nombre para ID único (CORREGIDO)"""
    # Quitar caracteres especiales SIN backslash en f-string
    nombre_limpio = re.sub(r'[^\w\s]', ' ', nombre_raw.lower())
    nombre_limpio = re.sub(r'\s+', ' ', nombre_limpio).strip()
    return nombre_limpio[:40]

def procesar_bloque(bloque, pagina_actual):
    """🎯 Procesamiento ultra estable de bloques (CORREGIDO)"""
    try:
        texto = bloque.text.strip()
        if len(texto) < 15 or '$' not in texto:
            return None
        
        precio = extraer_precio(texto)
        if precio == 0.0:
            return None
        
        # Líneas candidatas para nombre
        lineas = [l.strip() for l in texto.split('\n') if len(l.strip()) > 3]
        candidatos = []
        for linea in lineas:
            if '$' not in linea and not re.match(r'^\d+$', linea) and len(linea) > 5:
                candidatos.append(linea)
        
        if not candidatos:
            return None
        
        nombre_raw = max(candidatos, key=len)[:100].strip()
        
        # ID único CORREGIDO (sin backslash en f-string)
        nombre_limpio = limpiar_para_id(nombre_raw)
        id_unico = f"{nombre_limpio}_{int(precio)}_{pagina_actual}"
        
        if id_unico in productos_vistos:
            return None
        productos_vistos.add(id_unico)
        
        return {
            "nombre_producto": nombre_raw,
            "precio_unitario": precio,
            "pagina": pagina_actual
        }
    except:
        return None

# 🚀 PROCESO PRINCIPAL CON RECONEXIÓN
driver = None
for pagina_actual in range(1, MAX_PAGINAS + 1):
    try:
        # Reconexión automática
        if driver is None:
            print(f"\n🔄 Iniciando driver (página {pagina_actual})")
            driver = crear_driver()
            time.sleep(3)
        
        url = URL_BASE if pagina_actual == 1 else f"{URL_BASE}?currentPage={pagina_actual}"
        print(f"\n📄 Página {pagina_actual:2d}/15 ({pagina_actual/MAX_PAGINAS*100:>5.1f}%)", end="")
        
        driver.get(url)
        time.sleep(8)

        # Cookies
        try:
            WebDriverWait(driver, 5).until(
                EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Aceptar') or contains(text(), 'Acepto')]"))
            ).click()
            time.sleep(2)
        except:
            pass

        # SCROLL AGRESIVO
        for i in range(30):
            driver.execute_script(f"window.scrollTo(0, {i * 80});")
            time.sleep(0.6)
        time.sleep(4)

        # 🔍 Múltiples selectores para máxima captura
        selectores = [
            "div[class*='product-card']",
            "div[class*='product']",
            "div[class*='ProductCard']",
            "article[class*='product']",
            ".product-item",
            "[class*='product']"
        ]
        
        todos_bloques = []
        for selector in selectores:
            try:
                bloques = driver.find_elements(By.CSS_SELECTOR, selector)
                todos_bloques.extend([b for b in bloques if b not in todos_bloques])
            except:
                continue
        
        print(f" | {len(todos_bloques):>3} bloques")

        # Procesar
        nuevos = 0
        for bloque in todos_bloques:
            resultado = procesar_bloque(bloque, pagina_actual)
            if resultado:
                resultado.update({
                    "categoria_alimento": detectar_categoria(resultado["nombre_producto"]),
                    "marca": extraer_marca(resultado["nombre_producto"]),
                    "supermercado": "ACUENTA",
                    "fecha_scraping": time.strftime("%d/%m/%Y %H:%M")
                })
                datos_finales.append(resultado)
                nuevos += 1
        
        print(f"   ✅ +{nuevos:2d} productos | TOTAL: {len(datos_finales):4d}")

    except WebDriverException:
        print(" ❌ WebDriver falló - reconectando...")
        if driver:
            driver.quit()
        driver = None
        time.sleep(5)
        continue
    except Exception as e:
        print(f" ⚠️  Error: {str(e)[:50]}")
        continue

if driver:
    driver.quit()

# 🧮 ANÁLISIS ESTADÍSTICO PROFESIONAL
print(f"\n🔬 ANALIZANDO {len(datos_finales)} productos...")

precios_cat = defaultdict(list)
for p in datos_finales:
    precios_cat[p["categoria_alimento"]].append(p["precio_unitario"])

mapa_promedios = {}
for cat, precios in precios_cat.items():
    mapa_promedios[cat] = round(statistics.mean(precios), 0)

for p in datos_finales:
    p["precio_promedio_categoria"] = mapa_promedios[p["categoria_alimento"]]

# 📊 VISUALIZACIÓN PROFESIONAL
total_costo = sum(p["precio_unitario"] for p in datos_finales)
print("\n" + "="*140)
print(f"{'SUPERMERKADO':<15} | {'PRODUCTOS':<10} | {'COSTO TOTAL':<15} | {'PROMEDIO':<12} | {'CATEGORÍAS'}")
print("-"*140)
prom_general = statistics.mean([p['precio_unitario'] for p in datos_finales]) if datos_finales else 0
print(f"{'ACUENTA':<15} | {len(datos_finales):<10} | ${total_costo:>12,.0f} | ${prom_general:>9,.0f} | {len(precios_cat):<10}")
print("="*140)

print("\n🏆 TOP 15 PRODUCTOS (más caros):")
print("-"*140)
print(f"{'#':<4} | {'PRODUCTO':<50} | {'CATEGORÍA':<15} | {'PRECIO':<10} | {'PROM.CAT':<10} | {'MARCA':<12}")
print("-"*140)

top_productos = sorted(datos_finales, key=lambda x: x['precio_unitario'], reverse=True)[:15]
for i, p in enumerate(top_productos, 1):
    print(f"{i:<4} | {p['nombre_producto']:<50} | {p['categoria_alimento']:<15} | ${p['precio_unitario']:>8,.0f} | ${p['precio_promedio_categoria']:>8,.0f} | {p['marca']:<12}")

print("\n💰 PROMEDIOS POR CATEGORÍA (orden descendente):")
print("-"*80)
for cat in sorted(mapa_promedios.items(), key=lambda x: x[1], reverse=True):
    count = len([p for p in datos_finales if p["categoria_alimento"] == cat[0]])
    print(f"{cat[0]:<25} | ${cat[1]:>8,.0f} | ({count:>3} productos)")

# 💾 GUARDADO PROFESIONAL
with open("resultados_acuenta_PRO.json", "w", encoding="utf-8") as f:
    json.dump(datos_finales, f, ensure_ascii=False, indent=2)

print(f"\n🎉 ¡EXTRACCIÓN 100% EXITOSA!")
print(f"📦 {len(datos_finales)} productos únicos capturados")
print(f"💾 resultados_acuenta_PRO.json ✅")
print(f"⏰ {time.strftime('%d/%m/%Y %H:%M')}")
```

```python
import os
import time
import re
import json
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By

# 🧹 Limpieza de procesos previos para asegurar estabilidad
os.system("pkill -9 chrome")
os.system("pkill -9 chromedriver")

print("🧹 INICIANDO EXTRACCIÓN ULTRA ESTABLE - ACUENTA")

URL_BASE = "https://www.acuenta.cl/ca/despensa/05"
MAX_PAGINAS = 15
NOMBRE_RESPONSABLE = "Jorge Chávez"  # 👤 Responsable asignado
datos_finales = []
productos_vistos = set()

def crear_driver():
    """🔧 Configuración de Selenium para evitar bloqueos"""
    options = Options()
    options.binary_location = "/usr/bin/google-chrome"
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option('useAutomationExtension', False)
    options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
        "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined})"
    })
    return driver

# 🌈 CATEGORÍAS Y MARCAS
CATEGORIAS = {
    'arroz': 'Arroz', 'fideo': 'Pastas', 'pasta': 'Pastas', 'espagueti': 'Pastas', 'tallarines': 'Pastas',
    'aceite': 'Aceites', 'mayonesa': 'Salsas', 'ketchup': 'Salsas', 'salsa': 'Salsas',
    'atún': 'Conservas', 'jurel': 'Conservas', 'azúcar': 'Azúcares', 'harina': 'Harinas', 
    'porotos': 'Legumbres', 'lentejas': 'Legumbres', 'sal': 'Condimentos', 'sopa': 'Instantáneos'
}

def extraer_marca(nombre):
    marcas = ['LUCCHETTI','TUCAPEL','IANSA','CAROZZI','MIRAFLORES','HELLMANNS','LIDER','ACUENTA','SELECTA','LA ROMA','NETO','KRAFT','MAGGI','TRAVERSO','GOURMET','LOBOS','VIVO','BANQUETE']
    nombre_upper = nombre.upper()
    for marca in marcas:
        if marca in nombre_upper: return marca
    return "GENÉRICA"

def detectar_categoria(nombre):
    nombre_lower = nombre.lower()
    for palabra, cat in CATEGORIAS.items():
        if palabra in nombre_lower: return cat
    return "Despensa General"

def extraer_precio(texto):
    match = re.search(r'\$([0-9]{1,3}(?:\.[0-9]{3})*)', texto)
    if match:
        try: return float(match.group(1).replace('.', ''))
        except: pass
    return 0.0

def procesar_bloque(bloque, pagina_actual):
    try:
        texto = bloque.text.strip()
        if '$' not in texto: return None
        
        precio = extraer_precio(texto)
        lineas = [l.strip() for l in texto.split('\n') if len(l.strip()) > 3]
        if not lineas: return None
        
        nombre_raw = max(lineas, key=len)[:100].strip()
        
        # 📸 Extracción de URL de imagen
        try:
            img_element = bloque.find_element(By.TAG_NAME, "img")
            url_imagen = img_element.get_attribute("src")
        except:
            url_imagen = "Sin imagen"

        id_unico = f"{nombre_raw.lower()}_{int(precio)}"
        if id_unico in productos_vistos: return None
        productos_vistos.add(id_unico)
        
        return {
            "nombre_producto": nombre_raw,
            "precio_unitario": precio,
            "imagen": url_imagen,
            "responsable": NOMBRE_RESPONSABLE,
            "pagina": pagina_actual
        }
    except: return None

# 🚀 PROCESO DE SCRAPING
driver = None
for pagina_actual in range(1, MAX_PAGINAS + 1):
    try:
        if driver is None:
            driver = crear_driver()
        
        url = URL_BASE if pagina_actual == 1 else f"{URL_BASE}?currentPage={pagina_actual}"
        print(f"📄 Procesando Página {pagina_actual}/{MAX_PAGINAS}...", end="\r")
        
        driver.get(url)
        time.sleep(8) # Espera carga de elementos dinámicos

        # Scroll para activar Lazy Load
        for i in range(15):
            driver.execute_script(f"window.scrollTo(0, {i * 400});")
            time.sleep(0.3)

        bloques = driver.find_elements(By.CSS_SELECTOR, "div[class*='product-card'], .product-item")
        
        for bloque in bloques:
            res = procesar_bloque(bloque, pagina_actual)
            if res:
                res.update({
                    "categoria_alimento": detectar_categoria(res["nombre_producto"]),
                    "marca": extraer_marca(res["nombre_producto"]),
                    "supermercado": "ACUENTA",
                    "fecha_captura": time.strftime("%d/%m/%Y %H:%M")
                })
                datos_finales.append(res)
    except Exception as e:
        print(f"\n⚠️ Error en pág {pagina_actual}: {str(e)[:50]}")
        if driver: driver.quit()
        driver = None

if driver: driver.quit()

# ==============================================================================
# 📊 REPORTE DE CONSOLA ACTUALIZADO
# ==============================================================================
total_costo = sum(p["precio_unitario"] for p in datos_finales)

print("\n" + "="*145)
print(f"{'SUPERMERCADO':<15} | {'PRODUCTOS':<10} | {'COSTO TOTAL':<15} | {'RESPONSABLE':<20} | {'FECHA CAPTURA':<15} | {'ESTADO'}")
print("-" * 145)
print(f"{'ACUENTA':<15} | {len(datos_finales):<10} | ${total_costo:>12,.0f} | {NOMBRE_RESPONSABLE:<20} | {time.strftime('%d/%m/%Y'):<15} | {'FINALIZADO'}")
print("="*145)

# 💾 GUARDADO EN JSON
with open("resultados_acuenta_TODOS.json", "w", encoding="utf-8") as f:
    json.dump(datos_finales, f, ensure_ascii=False, indent=2)

print(f"\n✅ PROCESO COMPLETADO. Archivo 'resultados_acuenta_TODOS.json' generado para MongoDB.")
```

```python
import os
import time
import re
import json
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from pymongo import MongoClient

# 🧹 Limpieza de procesos previos para asegurar estabilidad
os.system("pkill -9 chrome")
os.system("pkill -9 chromedriver")

print("🧹 INICIANDO EXTRACCIÓN ULTRA ESTABLE - ACUENTA")

URL_BASE = "https://www.acuenta.cl/ca/despensa/05"
MAX_PAGINAS = 15
NOMBRE_RESPONSABLE = "Jorge Chávez"  # 👤 Responsable asignado
datos_finales = []
productos_vistos = set()

def crear_driver():
    """🔧 Configuración de Selenium para evitar bloqueos"""
    options = Options()
    options.binary_location = "/usr/bin/google-chrome"
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option('useAutomationExtension', False)
    options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
        "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined})"
    })
    return driver

# 🌈 CATEGORÍAS Y MARCAS
CATEGORIAS = {
    'arroz': 'Arroz', 'fideo': 'Pastas', 'pasta': 'Pastas', 'espagueti': 'Pastas', 'tallarines': 'Pastas',
    'aceite': 'Aceites', 'mayonesa': 'Salsas', 'ketchup': 'Salsas', 'salsa': 'Salsas',
    'atún': 'Conservas', 'jurel': 'Conservas', 'azúcar': 'Azúcares', 'harina': 'Harinas', 
    'porotos': 'Legumbres', 'lentejas': 'Legumbres', 'sal': 'Condimentos', 'sopa': 'Instantáneos'
}

def extraer_marca(nombre):
    marcas = ['LUCCHETTI','TUCAPEL','IANSA','CAROZZI','MIRAFLORES','HELLMANNS','LIDER','ACUENTA','SELECTA','LA ROMA','NETO','KRAFT','MAGGI','TRAVERSO','GOURMET','LOBOS','VIVO','BANQUETE']
    nombre_upper = nombre.upper()
    for marca in marcas:
        if marca in nombre_upper: return marca
    return "GENÉRICA"

def detectar_categoria(nombre):
    nombre_lower = nombre.lower()
    for palabra, cat in CATEGORIAS.items():
        if palabra in nombre_lower: return cat
    return "Despensa General"

def extraer_precio(texto):
    match = re.search(r'\$([0-9]{1,3}(?:\.[0-9]{3})*)', texto)
    if match:
        try: return float(match.group(1).replace('.', ''))
        except: pass
    return 0.0

def procesar_bloque(bloque, pagina_actual):
    try:
        texto = bloque.text.strip()
        if '$' not in texto: return None
        
        precio = extraer_precio(texto)
        lineas = [l.strip() for l in texto.split('\n') if len(l.strip()) > 3]
        if not lineas: return None
        
        nombre_raw = max(lineas, key=len)[:100].strip()
        
        try:
            img_element = bloque.find_element(By.TAG_NAME, "img")
            url_imagen = img_element.get_attribute("src")
        except:
            url_imagen = "Sin imagen"

        id_unico = f"{nombre_raw.lower()}_{int(precio)}"
        if id_unico in productos_vistos: return None
        productos_vistos.add(id_unico)
        
        return {
            "nombre_producto": nombre_raw,
            "precio_unitario": precio,
            "imagen": url_imagen,
            "responsable": NOMBRE_RESPONSABLE,
            "pagina": pagina_actual
        }
    except: return None

# 🚀 PROCESO DE SCRAPING
driver = None
for pagina_actual in range(1, MAX_PAGINAS + 1):
    try:
        if driver is None:
            driver = crear_driver()
        
        url = URL_BASE if pagina_actual == 1 else f"{URL_BASE}?currentPage={pagina_actual}"
        print(f"📄 Procesando Página {pagina_actual}/{MAX_PAGINAS}...", end="\r")
        
        driver.get(url)
        time.sleep(8) 

        for i in range(15):
            driver.execute_script(f"window.scrollTo(0, {i * 400});")
            time.sleep(0.3)

        bloques = driver.find_elements(By.CSS_SELECTOR, "div[class*='product-card'], .product-item")
        
        for bloque in bloques:
            res = procesar_bloque(bloque, pagina_actual)
            if res:
                res.update({
                    "categoria_alimento": detectar_categoria(res["nombre_producto"]),
                    "marca": extraer_marca(res["nombre_producto"]),
                    "supermercado": "ACUENTA",
                    "fecha_captura": time.strftime("%d/%m/%Y %H:%M")
                })
                datos_finales.append(res)
    except Exception as e:
        print(f"\n⚠️ Error en pág {pagina_actual}: {str(e)[:50]}")
        if driver: driver.quit()
        driver = None

if driver: driver.quit()

# 📊 REPORTE DE CONSOLA
total_costo = sum(p["precio_unitario"] for p in datos_finales)
print("\n" + "="*145)
print(f"{'SUPERMERCADO':<15} | {'PRODUCTOS':<10} | {'COSTO TOTAL':<15} | {'RESPONSABLE':<20} | {'FECHA CAPTURA':<15} | {'ESTADO'}")
print("-" * 145)
print(f"{'ACUENTA':<15} | {len(datos_finales):<10} | ${total_costo:>12,.0f} | {NOMBRE_RESPONSABLE:<20} | {time.strftime('%d/%m/%Y'):<15} | {'FINALIZADO'}")
print("="*145)

# 💾 GUARDADO EN JSON (Local)
with open("resultados_acuenta_TODOS.json", "w", encoding="utf-8") as f:
    json.dump(datos_finales, f, ensure_ascii=False, indent=2)

# ==============================================================================
# 📦 ENVÍO A MONGODB (Integrado)
# ==============================================================================
try:
    print("\n🔗 Conectando a MongoDB...")
    # Usa 'localhost' si corres el script manual, o 'mongodb' si es desde Docker
    client = MongoClient('mongodb://databse:27017/') 
    db = client['Canasta_dbacuenta'] 
    coleccion = db['Retail_A']

    if datos_finales:
        print(f"📤 Subiendo {len(datos_finales)} productos a la colección 'Retail_A'...")
        coleccion.insert_many(datos_finales)
        print("✅ ¡Carga exitosa! Ya puedes revisar Mongo Express (Puerto 8081).")
    else:
        print("⚠️ No se encontraron productos para subir.")

except Exception as e:
    print(f"❌ Error al conectar con MongoDB: {e}")

print(f"\n✅ PROCESO COMPLETADO.")
```
